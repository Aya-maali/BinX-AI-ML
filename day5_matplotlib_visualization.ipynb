# Day 5 - Data Visualization with Matplotlib

In this notebook, we learn how to visualize data using Matplotlib.
We will create different types of plots to better understand and communicate data.

## Why Visualization Matters

Visualization helps us understand data by revealing patterns, trends, relationships, and outliers that are difficult to notice in tables.
It is an important tool for both data exploration and presenting results.

## Matplotlib Basics

Matplotlib is the main plotting library in Python.
The basic workflow is:
1. Create a plot.
2. Add labels.
3. Add a title.
4. Display the figure.

```python
import matplotlib.pylab as plt

x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

plt.plot(x, y) # for plotting the line graph
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Simple Line Plot')
plt.show()
```

## The Four Core Plot Types

## 1- Line Plot

A line plot is used to show trends or changes over time or another continuous variable.

```python
months =[1, 2, 3, 4, 5, 6]
sales = [100, 200, 300, 400, 500, 600]

plt.plot(months,sales)
plt.xlabel('Months')
plt.ylabel('Sales')
plt.title('Sales Data')
plt.show()
```

## 2- Scatter Plot

A scatter plot shows the relationship between two numerical variables.

```python
age = [20, 25, 30, 35, 40, 45]
salary = [30000, 40000, 50000, 60000, 70000, 80000]

plt.scatter(age, salary)
plt.xlabel('Age')
plt.ylabel('Salary')
plt.title('Age vs Salary')
plt.show()
```

## 3- Bar Chart

A bar chart compares values across different categories.

```python
majors = ['Computer Science', 'Mathematics', 'Physics', 'Chemistry', 'Biology']
students = [100, 80, 60, 40, 50]

plt.bar(majors, students)
plt.xlabel('Majors')    
plt.ylabel('Number of Students')
plt.title('Number of Students in Different Majors')
plt.show()
```

## 4- Histogram

A histogram shows the distribution of a numerical variable.

```python
grades = [60, 70, 75, 80, 82, 85, 90, 92, 95, 98]

plt.hist(grades)

plt.xlabel("Grade")
plt.ylabel("Frequency")
plt.title("Grade Distribution")

plt.show()
```

## Subplots

Subplots allow us to display multiple charts in the same figure for easy comparison.

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

grades = [60, 70, 75, 80, 82, 85, 90, 92, 95, 98]
axes[0].hist(grades)
axes[0].set_title("Grade Distribution")

ages = [20, 25, 30, 35, 40, 45]
salaries = [30000, 40000, 50000, 60000, 70000, 80000]
axes[1].scatter(ages, salaries)
axes[1].set_title("Age vs Salary")

plt.tight_layout() # Adjusts the spacing between subplots for better layout
plt.show()
```

# Hands-On Lab - Week 1 Integrated Mini-Notebook

In this notebook, we combine Pandas, NumPy, and Matplotlib to load, clean, analyze, and visualize a dataset.

## Task 1: Load and Clean the Dataset

Load the CSV dataset using Pandas and check for missing values.
If any missing values exist, clean the dataset by removing or filling them.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

#load the dataset
df = pd.read_csv("students.csv")

# Display the first few rows of the dataset
print(df.head())

#check for missing values
df.isnull().sum()

#if there are missing values
df = df.dropna()  # Drop rows with missing values
```

## Task 2: NumPy Computation

Use NumPy to calculate a summary statistic from the dataset.

```python
average_grade = np.mean(df['Grade'])
print(f"Average Grade: {average_grade}")

maximum_grade = np.max(df['Grade'])
print(f"Maximum Grade: {maximum_grade}")

minimum_grade = np.min(df['Grade'])
print(f"Minimum Grade: {minimum_grade}")
```

## Task 3: Data Visualization

Create three labeled plots to explore the dataset.

```python
# 1- Bar chart for the number of students in each major

major_counts = df['Major'].value_counts()
plt.figure(figsize=(6,4))
plt.bar(major_counts.index, major_counts.values)
plt.xlabel('Majors')
plt.ylabel('Number of Students')
plt.title('Number of Students in Each Major')
plt.show()
```

```python
# 2- Histogram for the distribution of grades

plt.figure(figsize=(6,4))
plt.hist(df['Grade'])
plt.xlabel('Grades')
plt.ylabel('Frequency')
plt.title('Distribution of Grades')
plt.show()
```

```python
# 3- Scatter plot for Age vs Grade

plt.figure(figsize=(6,4))
plt.scatter(df['Age'], df['Grade'])
plt.xlabel('Age')
plt.ylabel('Grade')
plt.title('Age vs Grade')
plt.show()
```
