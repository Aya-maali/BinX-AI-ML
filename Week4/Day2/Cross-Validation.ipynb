# Topic 1: Why Cross-Validation?

A single train/validation split can sometimes give a misleading result.

The validation set might contain an unusual group of samples, so the model's score can depend too much on the specific split.

Cross-validation solves this problem by creating multiple validation sets.

Instead of evaluating the model on only one validation set, we evaluate it several times using different parts of the training data.

This gives us a more stable and reliable estimate of model performance.

```python
from sklearn.model_selection import cross_val_score
from sklearn.neighbors import KNeighborsClassifier

# Create the KNN model
model = KNeighborsClassifier(n_neighbors=3)

# Perform 5-fold cross-validation
scores = cross_val_score(
    model,
    X_train,
    y_train,
    cv=5,
    scoring="accuracy"
)

print("Cross-validation scores:")
print(scores)
```

```python
import numpy as np

mean_score = np.mean(scores)
std_score = np.std(scores)

print("Mean Accuracy:", mean_score)
print("Standard Deviation:", std_score)

print(f"Result: {mean_score:.2f} ± {std_score:.2f}")
```

# Topic 2: How K-Fold Cross-Validation Works

In K-Fold Cross-Validation, the training data is divided into `k` equal parts called folds.

For example, with 5-fold cross-validation:

- Fold 1 is used for validation and the other 4 folds are used for training.
- Fold 2 is used for validation and the other 4 folds are used for training.
- Fold 3 is used for validation and the other 4 folds are used for training.
- Fold 4 is used for validation and the other 4 folds are used for training.
- Fold 5 is used for validation and the other 4 folds are used for training.

The model is trained and evaluated 5 times.

Each fold gets a chance to be the validation set.

The final performance is calculated from all validation scores.

```python
from sklearn.model_selection import KFold

kf = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

print("Number of folds:", kf.get_n_splits(X_train))
```

# Topic 3: Using cross_val_score

`cross_val_score` is a Scikit-learn function that makes cross-validation easier.

It:

1. Splits the data into folds.
2. Trains the model on the training folds.
3. Evaluates the model on the validation fold.
4. Repeats the process for every fold.
5. Returns one score for each fold.

We can use these scores to understand how consistently the model performs.

```python
from sklearn.model_selection import cross_val_score
from sklearn.neighbors import KNeighborsClassifier

model = KNeighborsClassifier(n_neighbors=3)

scores = cross_val_score(
    model,
    X_train,
    y_train,
    cv=kf,
    scoring="accuracy"
)

print("Scores for each fold:")
for i, score in enumerate(scores, start=1):
    print(f"Fold {i}: {score:.2f}")
```

# Topic 4: Mean and Standard Deviation

Cross-validation produces multiple scores.

The **mean** represents the average performance across all folds.

The **standard deviation** shows how much the scores vary between folds.

A small standard deviation means that the model performs similarly across different folds.

A large standard deviation means that the model's performance changes significantly depending on the data used for validation.

Therefore, we should report both the mean and standard deviation.

```python
import numpy as np

mean_score = np.mean(scores)
std_score = np.std(scores)

print("Mean Accuracy:", mean_score)
print("Standard Deviation:", std_score)

print(f"\nResult: {mean_score:.2f} ± {std_score:.2f}")
```

# Topic 5: Stratified K-Fold Cross-Validation

For classification problems, we should consider the distribution of the target classes.

For example, our target variable `Passed` contains two classes:

- `0` = Did not pass
- `1` = Passed

Stratified K-Fold tries to preserve the same class distribution in every fold.

This is useful because every validation fold should contain a representative mixture of the classes.

For classification problems, Stratified K-Fold is usually preferred over regular K-Fold.

```python
from sklearn.model_selection import StratifiedKFold

stratified_kfold = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

print("Number of folds:", stratified_kfold.get_n_splits(X_train, y_train))
```

# Topic 6: Cross-Validation with Stratified K-Fold

We can combine `StratifiedKFold` with `cross_val_score`.

The model will be trained and evaluated multiple times.

Each fold will maintain a similar distribution of the target classes.

This makes the evaluation more appropriate for classification problems.

```python
from sklearn.model_selection import cross_val_score
from sklearn.neighbors import KNeighborsClassifier

model = KNeighborsClassifier(n_neighbors=3)

stratified_scores = cross_val_score(
    model,
    X_train,
    y_train,
    cv=stratified_kfold,
    scoring="accuracy"
)

print("Stratified K-Fold scores:")

for i, score in enumerate(stratified_scores, start=1):
    print(f"Fold {i}: {score:.2f}")
```

# Topic 7: Report the Cross-Validation Result

After running Stratified K-Fold Cross-Validation, we calculate:

- The mean accuracy.
- The standard deviation.

The mean tells us the average performance.

The standard deviation tells us how stable the performance is across the folds.

We can report the result as:

**Mean Accuracy ± Standard Deviation**

```python
mean_accuracy = np.mean(stratified_scores)
std_accuracy = np.std(stratified_scores)

print(f"Mean Accuracy: {mean_accuracy:.2f}")
print(f"Standard Deviation: {std_accuracy:.2f}")
print(f"Final Result: {mean_accuracy:.2f} ± {std_accuracy:.2f}")
```

# Hands-On Lab: Cross-Validating a Model

# Task Step 1: Prepare the Dataset

We will use the Week 3 student dataset.

The target variable is `Passed`.

The input features are:

- `Age`
- `Income`
- `StudyHours`
- `City`

Because `City` is categorical, it needs to be converted into numerical features before training the model.

We will also keep a separate test set that will not be used during cross-validation.

```python
import pandas as pd
from sklearn.model_selection import train_test_split

# Load the dataset
df = pd.read_csv("students.csv")

# Separate features and target
X = df.drop("Passed", axis=1)
y = df["Passed"]

# Encode the categorical City column
X = pd.get_dummies(
    X,
    columns=["City"],
    dtype=int
)

# Keep the test set untouched
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training set:", X_train.shape)
print("Test set:", X_test.shape)
```

```python
X = df.drop("Passed", axis=1)
y = df["Passed"]

X = pd.get_dummies(
    X,
    columns=["City"],
    dtype=int
)
```

# Task Step 2: Apply Stratified 5-Fold Cross-Validation

We will use Stratified 5-Fold Cross-Validation.

The training data will be divided into 5 folds.

Each fold will be used once as validation data while the other folds are used for training.

Stratification keeps the distribution of the `Passed` classes similar across the folds.

The test set remains untouched.

```python
from sklearn.model_selection import StratifiedKFold
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import cross_val_score

# Create Stratified 5-Fold
skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# Create the model
model = KNeighborsClassifier(n_neighbors=3)

# Perform cross-validation
cv_scores = cross_val_score(
    model,
    X_train,
    y_train,
    cv=skf,
    scoring="accuracy"
)

print("Cross-validation scores:")

for i, score in enumerate(cv_scores, start=1):
    print(f"Fold {i}: {score:.2f}")
```

# Task Step 3: Calculate Mean and Standard Deviation

We will calculate the mean and standard deviation of the five cross-validation scores.

The mean represents the average model performance across the folds.

The standard deviation represents how much the performance changes between folds.

A smaller standard deviation indicates more consistent performance.

```python
import numpy as np

mean_score = np.mean(cv_scores)
std_score = np.std(cv_scores)

print(f"Mean Accuracy: {mean_score:.2f}")
print(f"Standard Deviation: {std_score:.2f}")

print(f"\nCross-Validation Result: {mean_score:.2f} ± {std_score:.2f}")
```

# Task Step 4: Compare Different K Values

We will compare different values of `n_neighbors`.

For each value, we will perform Stratified 5-Fold Cross-Validation.

The mean cross-validation accuracy will be used to compare the models.

The value with the highest mean accuracy will be selected as the best value.

Unlike Day 1, we are using multiple validation folds instead of relying on one validation split.

```python
k_values = [1, 3, 5]

results = []

for k in k_values:

    model = KNeighborsClassifier(
        n_neighbors=k
    )

    scores = cross_val_score(
        model,
        X_train,
        y_train,
        cv=skf,
        scoring="accuracy"
    )

    mean_score = np.mean(scores)
    std_score = np.std(scores)

    results.append({
        "k": k,
        "mean_accuracy": mean_score,
        "std_accuracy": std_score
    })

    print(
        f"k={k} | "
        f"Mean Accuracy={mean_score:.2f} | "
        f"Std={std_score:.2f}"
    )
```

# Task Step 5: Select the Best Hyperparameter

We will select the value of `k` that achieves the highest mean cross-validation accuracy.

The cross-validation results are used for model selection.

The test set is still untouched.

After selecting the best value, the final model can be trained and evaluated on the held-out test set.

```python
best_result = max(
    results,
    key=lambda x: x["mean_accuracy"]
)

best_k = best_result["k"]

print("Best k:", best_k)
print("Best Mean Accuracy:", best_result["mean_accuracy"])
print("Standard Deviation:", best_result["std_accuracy"])
```

# Task Step 6: Final Evaluation on the Test Set

After selecting the best `k` using cross-validation, we train the final model.

The test set is used only once for the final evaluation.

This gives us a final estimate of performance on unseen data.

The test set was not used to choose the best hyperparameter.

```python
from sklearn.metrics import accuracy_score

# Create final model
final_model = KNeighborsClassifier(
    n_neighbors=best_k
)

# Train on the complete training set
final_model.fit(
    X_train,
    y_train
)

# Predict the test set
y_test_pred = final_model.predict(X_test)

# Calculate test accuracy
test_accuracy = accuracy_score(
    y_test,
    y_test_pred
)

print("Best k:", best_k)
print("Final Test Accuracy:", test_accuracy)
```

# Task Step 7: Final Analysis

Cross-validation provided multiple performance scores instead of relying on a single validation split.

The mean accuracy represents the average performance across the folds.

The standard deviation shows how stable the model performance was across different folds.

Stratified K-Fold was used because this is a classification problem and it helps maintain a similar class distribution across the folds.

The test set was kept separate and was used only once for the final evaluation.

This workflow provides a more reliable estimate of model performance than using a single validation split.
