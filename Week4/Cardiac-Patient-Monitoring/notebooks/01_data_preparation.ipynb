## Project & Dataset Selection

**Chosen project topic:** *Cardiac Patient Monitoring System*.

**Chosen dataset:** the **UCI Heart Disease dataset** (Cleveland database, UCI repository ID 45). It contains **303 patient records** described by **13 clinical features** (age, sex, chest pain type, resting blood pressure, cholesterol, fasting blood sugar, resting ECG, max heart rate, exercise-induced angina, ST depression, slope, number of major vessels, thalassemia result), with a target column indicating heart disease presence/severity (0 = no disease, 1-4 = increasing severity).

**Why this dataset was chosen:**
- It is a **public, non-identifiable** dataset (no names, dates, or medical-record numbers), which matches the project's out-of-scope constraint on identifiable patient data.
- It is **curriculum-aligned**: it needs only Pandas/NumPy for cleaning, Matplotlib for EDA, and Scikit-learn for classification, feature engineering/pipelines, and clustering/PCA — no deep learning or external infrastructure.
- It is a well-documented, classic classification dataset, so results can be reasoned about and sanity-checked against known findings in the literature.
- Its feature set is small and mostly numeric/ordinal, which keeps the feature-engineering and pipeline work (Milestone M5) and the unsupervised analysis (Milestone M6) meaningful and easy to interpret.

**How it was obtained:** the dataset is loaded programmatically through the `ucimlrepo` Python package, using `fetch_ucirepo(id=45)`, which pulls the official UCI Machine Learning Repository entry for *Heart Disease* directly. This keeps data acquisition reproducible — no manual file downloads are needed to re-run this project.

### 4. Handling Missing Values

`ca` and `thal` contain a small number of missing values. Rather than dropping or imputing them here, imputation is deferred to the `SimpleImputer` step inside the Scikit-learn `Pipeline` built in `04_feature_engineering_pipeline.ipynb`, so it is learned only from training data (avoiding data leakage) and stays fully reproducible.

```python
from sklearn.impute import SimpleImputer

# Create an imputer for categorical features
categorical_imputer = SimpleImputer(strategy="most_frequent")

# Check the most frequent values
for column in ["ca", "thal"]:
    most_frequent = df[column].mode()[0]
    print(f"{column}: {most_frequent}")
```
