# Probability Basics

Probability measures how likely an event is to happen.

A probability is always between 0 and 1.

- 0 means impossible.
- 1 means certain.

Formula:

Probability = Favorable Outcomes / Total Outcomes

Probability is widely used in machine learning to estimate the likelihood of predictions.

```python
# Probability of getting number 3 on a die

favorable = 1
total = 6

probability = favorable / total

print("Probability:", probability)
```

```python
# A bag contains:
# 4 Red balls
# 3 Blue balls
# 3 Green balls

red = 4
blue = 3
green = 3

total = red + blue + green

print("Probability of Red:", red / total)
```

# Complement Rule

The complement of an event is the probability that the event does not happen.

Formula:

P(A') = 1 - P(A)

Where:

- P(A) is the probability that the event happens.
- P(A') is the probability that the event does not happen.

The sum of an event and its complement is always equal to 1.

```python
# Probability of not getting 6

p_six = 1 / 6

p_not_six = 1 - p_six

print("P(Not 6):", p_not_six)
```

```python
# Student passing probability

p_pass = 0.8

p_fail = 1 - p_pass

print("Probability of Failure:", p_fail)
```

```python
p_rain = 0.6

print("Probability of No Rain:", 1 - p_rain)
```

# Addition Rule

The addition rule is used to calculate the probability of event A or event B.

For mutually exclusive events:

P(A ∪ B) = P(A) + P(B)

For overlapping events:

P(A ∪ B) = P(A) + P(B) - P(A ∩ B)

```python
# Rolling a die

p_one = 1 / 6
p_two = 1 / 6

print("Probability:", p_one + p_two)
```

```python
p_programming = 0.6
p_math = 0.4
p_both = 0.2

probability = p_programming + p_math - p_both

print(probability)
```

# Multiplication Rule

The multiplication rule is used to calculate the probability that two events happen together.

For independent events:

P(A ∩ B) = P(A) × P(B)

For dependent events:

P(A ∩ B) = P(A) × P(B|A)

Independent events do not affect each other.

Dependent events change the probability of the next event.

```python
# Independent: Coin and Die

p_heads = 1 / 2
p_six = 1 / 6

probability = p_heads * p_six

print("Probability:", probability)
```

```python
#Dependent:
# Bag with 3 red and 2 blue balls
# Without replacement

p_red = 3 / 5
p_blue_after_red = 2 / 4

probability = p_red * p_blue_after_red

print("Probability:", probability)
```

# Conditional Probability

Conditional probability measures the probability of an event given that another event has already occurred.

Formula:

P(A|B) = P(A ∩ B) / P(B)

Conditional probability is useful when one event changes the likelihood of another event.

```python
# Conditional Probability

p_a_and_b = 0.3
p_b = 0.6

conditional_probability = p_a_and_b / p_b

print("P(A|B):", conditional_probability)
```

```python
p_pass_and_study = 0.45
p_study = 0.60

print("Conditional Probability:", p_pass_and_study / p_study)
```

# Bayes' Theorem

Bayes' Theorem updates the probability of an event using new evidence.

Formula:

P(A|B) = (P(B|A) × P(A)) / P(B)

Where:

- P(A) is the prior probability.
- P(B|A) is the likelihood.
- P(B) is the probability of the evidence.
- P(A|B) is the updated probability.

Bayes' Theorem is widely used in machine learning, especially in Naive Bayes classification.

```python
p_a = 0.4
p_b_given_a = 0.8
p_b = 0.5

bayes = (p_b_given_a * p_a) / p_b

print("P(A|B):", bayes)
```

```python
p_disease = 0.02
p_positive_given_disease = 0.95
p_positive = 0.05

result = (p_positive_given_disease * p_disease) / p_positive

print(result)
```

## Probability Distributions

# 1- Normal Distribution (Bell Curve)

The normal distribution is a symmetric, bell-shaped distribution.

Characteristics:

- Mean = Median = Mode
- Most values are close to the mean.
- Extreme values are less common.

Many real-world datasets approximately follow a normal distribution.

```python
import numpy as np

data = np.random.normal(loc=70, scale=10, size=1000)

print("Mean:", np.mean(data))
print("Standard Deviation:", np.std(data))
```

# 2- Binomial Distribution

The binomial distribution models the number of successes in a fixed number of independent trials.

Requirements:

- Fixed number of trials.
- Two possible outcomes.
- Independent trials.
- Constant probability of success.

Examples include coin tosses, pass/fail tests, and yes/no questions.

```python
from scipy.stats import binom

# Probability of getting exactly 5 heads
probability = binom.pmf(5, n=10, p=0.5)

print(probability)
```

# Step 1: Simulating 10,000 Coin Flips

In this experiment, we simulate flipping a fair coin 10,000 times using NumPy.

A fair coin has two equally likely outcomes:
- Heads (1)
- Tails (0)

According to probability theory, as the number of flips increases, the proportion of heads should approach 0.5.

This demonstrates the **Law of Large Numbers**, where experimental probability gets closer to theoretical probability as the number of trials increases.

```python
import numpy as np

# Simulate 10,000 coin flips
coin_flips = np.random.choice([0, 1], size=10000)

heads = np.sum(coin_flips)
tails = len(coin_flips) - heads

print("Heads:", heads)
print("Tails:", tails)

print("Proportion of Heads:", heads / len(coin_flips))
print("Proportion of Tails:", tails / len(coin_flips))
```

# Step 3: Conditional Probability

Suppose a bag contains:

- 3 Red balls
- 2 Blue balls

A ball is drawn without replacement.

We want to calculate the probability of drawing a Blue ball after a Red ball has already been drawn.

The theoretical probability is:

P(Blue | Red) = 2 / 4 = 0.5

We verify this result using a simulation with many repeated experiments.

```python
import numpy as np

trials = 10000
success = 0

for _ in range(trials):
    bag = ["R", "R", "R", "B", "B"]

    first = np.random.choice(bag)
    bag.remove(first)

    second = np.random.choice(bag)

    if first == "R" and second == "B":
        success += 1

print("Simulated Probability:", success / trials)
print("Theoretical Probability:", (3/5) * (2/4))
```
