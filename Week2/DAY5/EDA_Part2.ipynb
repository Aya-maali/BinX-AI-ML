# Bivariate Analysis

Bivariate analysis examines the relationship between two variables.

It helps answer questions such as:

- Do two variables have a relationship?
- Does one variable increase as another increases?
- Are there differences between categories?

Common visualizations include scatter plots and grouped box plots.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load Titanic dataset
df = sns.load_dataset("titanic")


df.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

sns.scatterplot(data=df, x="age", y="fare")

plt.show()
```

## Scatter Plot

A scatter plot shows the relationship between two numerical variables.

Each point represents one observation.

Scatter plots help identify:

- Positive relationships
- Negative relationships
- Clusters
- Outliers

```python
sns.scatterplot(
    data=df,
    x="age",
    y="fare"
)

plt.show()
```

## Grouped Box Plot

Grouped box plots compare the distribution of a numerical variable across different categories.

They help compare medians, spread, and outliers between groups.

```python
sns.boxplot(
    data=df,
    x="class",
    y="fare"
)

plt.show()
```

# Correlation

Correlation measures the strength and direction of the relationship between two numerical variables.

Correlation values range from:

- -1 : Strong negative correlation
- 0 : No linear correlation
- +1 : Strong positive correlation

```python
corr = df.corr(numeric_only=True)

corr
```

# Correlation Heatmap

A heatmap visualizes the correlation matrix using colors.

It helps quickly identify strong positive and negative relationships between numerical variables.

```python
corr = df.corr(numeric_only=True)

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.show()
```

# Correlation Is Not Causation

A strong correlation between two variables does not mean that one variable causes the other.

Correlation identifies relationships, but it cannot prove cause and effect.

# Pair Plot

A pair plot displays relationships between all numerical variables in a dataset.

It combines scatter plots and distributions into a single visualization, making it useful for quickly exploring relationships.

```python
sns.pairplot(df)

plt.show()
```

# Step 1: Bivariate Analysis

In this step, I explored the relationships between important variables using scatter plots and grouped box plots.

Scatter plots help identify relationships between numerical variables, while grouped box plots compare numerical distributions across different categories.

```python
# Scatter Plot

plt.figure(figsize=(6,4))

sns.scatterplot(
    data=df,
    x="age",
    y="fare"
)

plt.title("Age vs Fare")
plt.show()
```

```python
#Grouped Box Plot

plt.figure(figsize=(6,4))

sns.boxplot(
    data=df,
    x="class",
    y="fare"
)

plt.title("Fare by Passenger Class")
plt.show()
```

### Observation

The scatter plot shows that there is no strong linear relationship between Age and Fare.

The grouped box plot shows that First Class passengers generally paid higher fares than passengers in Second and Third Class.

# Step 2: Correlation Analysis

A correlation matrix measures the strength of relationships between numerical variables.

A heatmap provides a visual representation of these relationships.

```python
corr = df.corr(numeric_only=True)

corr
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```

### Observation

The heatmap shows both positive and negative correlations.

Most variables have weak correlations, while a few variables have moderate relationships.

# Step 3: Strongest Relationships

Based on the correlation matrix, several variables show stronger relationships than others.

Examples include:

- Passenger class and fare.
- Passenger class and survival.
- Fare and survival.

These relationships may be useful features when building a machine learning model.
