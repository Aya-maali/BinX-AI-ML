- Dot Product
- Matrix Multiplication

# Vectors

A vector is an ordered list of numbers.

In Machine Learning, each data sample is represented as a vector.

Example:

[22, 3.6, 90]

This vector can represent:

- Age
- GPA
- Exam Score

```python
import numpy as np

student = np.array([22, 3.6, 90])

print(student)
print(student.shape)
```

# Matrices

A matrix is a two-dimensional array.

Rows represent data samples.

Columns represent features.

Example:

[
 [22,3.6,90],
 [21,3.1,75],
 [23,3.9,98]
]

```python
import numpy as np

students = np.array([
    [22,3.6,90],
    [21,3.1,75],
    [23,3.9,98]
])

print(students)
print(students.shape)
print(students[0])
print(students[:,1])
```

# Dot Product

The dot product multiplies corresponding elements of two vectors and sums the results.

It is the core operation used by many machine learning models to make predictions.

Prediction = Dot Product(Features, Weights)

```python
import numpy as np

features = np.array([25,50000,3])

weights = np.array([0.1,0.0002,1.5])

prediction = np.dot(features,weights)

print(prediction)
```

# Matrix Multiplication

Matrix multiplication allows machine learning models to compute predictions for multiple samples simultaneously.

The shapes must be compatible.

Example:

Matrix (Samples × Features)

×

Weight Vector (Features)

=

Prediction Vector (Samples)

```python
X = np.array([
    [25,50000,3],
    [40,80000,10]
])
w = np.array([0.1,0.0002,1.5])

predictions = X @ w

print(predictions)
```

# Task 2: Compute the Dot Product

## Objective
Calculate the prediction for one sample using the dot product.

The feature vector is multiplied by the weight vector.

```python
import numpy as np

features = np.array([25, 50000, 3])

weights = np.array([0.1, 0.0002, 1.5])

manual_prediction = (
    25 * 0.1 +
    50000 * 0.0002 +
    3 * 1.5
)

print("Manual Prediction:", manual_prediction)

prediction = np.dot(features, weights)

print("NumPy Prediction:", prediction)
```

# Task 3: Matrix Multiplication

## Objective
Use matrix multiplication to generate predictions for all samples simultaneously.

```python
import numpy as np

X = np.array([
    [25, 50000, 3],
    [40, 80000, 10],
    [33, 62000, 5]
])

weights = np.array([0.1, 0.0002, 1.5])

predictions = X @ weights

print("Predictions:")
print(predictions)
```

## Objective
Create a shape mismatch error intentionally, then explain why it happened and how to fix it.

```python
import numpy as np

X = np.array([
    [25, 50000, 3],
    [40, 80000, 10],
    [33, 62000, 5]
])

wrong_weights = np.array([0.1, 0.0002])

try:
    result = X @ wrong_weights
    print(result)
except ValueError as e:
    print("Error:")
    print(e)
```
