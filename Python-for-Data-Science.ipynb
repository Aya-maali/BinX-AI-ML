# Python Data Types

This section demonstrates Python's basic data types used in Data Science.

```python
# Integer
age = 22

# Float
gpa = 3.85

# String
name = "Aya"

# Boolean
graduated = False

# List
courses = ["Python", "AI", "ML"]

# Tuple
location = ("Jenin", "Palestine")

# Dictionary
student = {
    "name": "Aya",
    "age": 22
}

# Set
numbers = {1, 2, 3, 4}

print(type(age))
print(type(gpa))
print(type(name))
print(type(graduated))
print(type(courses))
print(type(location))
print(type(student))
print(type(numbers))
```

```python
x = [1, 2, 3]
print(type(x))
```

## Control Flow

Examples of if ,elif ,else ,for ,while..

```python
# if/else statement

score = 85
if score >= 90:
    print("pass")
else:
    print("fail")

print("________________")

# for loop

fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
    
    
print("________________")

# while loop

count = 0
while count < 5:
    print(count)
    count += 1
```

## Functions

Functions help organize reusable code.

```python
# Function to calculate the square of a number

def square(num):
    return num * num #**2
print(square(5))

#function to greet a person
def greet(name):
    return f"Hello, {name}!"
print(greet("Aya"))

#multiplication function
def multiply(a, b):
    return a * b
result = multiply(4, 6)
print(result)
```

## List Comprehensions

List comprehensions provide a concise way to create lists.

```python
#normal for loop to create a list of squares

square =[]

for x in range(5):
    square.append(x**2)
print(square)

#list comprehension to create a list of squares

squares = [x**2 for x in range (7)]
print(squares)

#with conditional statement

evens =[]
for x in range(10):
    if x%2 ==0:
        evens.append(x)
print(evens)

#list comprehension with conditional statement
even=[x for x in range(10) if x %2 ==0]
print(even)

#Q : what the output of the following code?
numbers = [x for x in range(6)]
print(numbers)
#Answer: [0, 1, 2, 3, 4, 5]

```

## Object-Oriented Programming (OOP)

A simple class example with attributes and a method.


```python
# class for student information(name , age)
      
class Student:
    def __init__(self, name, age):#constructor method
        self.name = name 
        self.age = age 
    
    # method to display student information
    def display_info(self): 
        print(f"Name: {self.name}, Age: {self.age}")

Student1 = Student("Aya", 22)
Student1.display_info()

        
        
```
