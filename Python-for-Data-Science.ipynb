# Python Data Types

This section demonstrates Python's basic data types used in Data Science.

```python
# Integer
age = 22

# Float
gpa = 3.85

# String
name = "Aya"

# Boolean
graduated = False

# List
courses = ["Python", "AI", "ML"]

# Tuple
location = ("Jenin", "Palestine")

# Dictionary
student = {
    "name": "Aya",
    "age": 22
}

# Set
numbers = {1, 2, 3, 4}

print(type(age))
print(type(gpa))
print(type(name))
print(type(graduated))
print(type(courses))
print(type(location))
print(type(student))
print(type(numbers))
```

```python
x = [1, 2, 3]
print(type(x))
```

## Control Flow

Examples of if ,elif ,else ,for ,while..

```python
# if/else statement

score = 85
if score >= 90:
    print("pass")
else:
    print("fail")

print("________________")

# for loop

fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
    
    
print("________________")

# while loop

count = 0
while count < 5:
    print(count)
    count += 1
```

## Functions

Functions help organize reusable code.

```python
# Function to calculate the square of a number

def square(num):
    return num * num #**2
print(square(5))

#function to greet a person
def greet(name):
    return f"Hello, {name}!"
print(greet("Aya"))

#multiplication function
def multiply(a, b):
    return a * b
result = multiply(4, 6)
print(result)
```
