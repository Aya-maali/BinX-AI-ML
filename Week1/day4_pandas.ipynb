# Pandas Data Structures

Pandas provides two main data structures for working with data:

### Series
A Series is a one-dimensional labeled array that stores a single column of data.

### DataFrame
A DataFrame is a two-dimensional table with rows and columns, similar to an Excel spreadsheet. It is made up of multiple Series objects.

```python
import pandas as pd

#Example 1: Create a Series from a list
ages = pd.Series([20, 22, 19, 21], name="Age")

print(ages)

#Example 2: Create a DataFrame from a dictionary
students = {
    "Name": ["Aya", "Ali", "Sara"],
    "Age": [22, 20, 23]
}

df = pd.DataFrame(students)

df
```

## Selecting Columns

We can select one or more columns from a DataFrame.

```python
import pandas as pd
students ={
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age": [20, 21, 19, 22],
    "Grade": ["A", "B", "C", "A"],
    "Major": ["Computer Science", "Mathematics", "Physics", "Engineering"]
    
}

df = pd.DataFrame(students)
df # to display the DataFrame 
df["Name"] # to display the Name column
df[["Name", "Age"]] # to display the Name and Age columns
df.iloc[0] # to display the first row
df.loc[1,"Age"] = 22 # to update the Age of the second student (Bob) to 22
df.iloc[1] # to display the updated second row
df["city"] = ["New York", "Los Angeles", "Chicago", "Houston"] # to add a new column 'city'
df["Passed"] = df["Grade"] <= "D" # to add a new column 'Passed' based on the Grade
df = df.drop(columns = ["Passed"]) # to drop the 'Passed' column
df
```

## Reading CSV Files

A CSV (Comma-Separated Values) file stores tabular data in plain text.
Pandas provides the `read_csv()` function to load a CSV file into a DataFrame.

```python
import pandas as pd
df = pd.read_csv("students.csv") # to read a CSV file into a DataFrame
df # to display the DataFrame
```

## Viewing the Dataset

Before analyzing data, it is important to inspect its contents.
Pandas provides functions such as `head()` and `tail()` to preview the dataset.

```python
df.head() # to display the first 5 rows of the DataFrame
df.head(3) # to display the first 3 rows of the DataFrame
df.tail() # to display the last 5 rows of the DataFrame
df.tail(2) # to display the last 2 rows of the DataFrameq
```

## Cleaning Data

Real-world datasets often contain missing values.
Before analyzing the data, we should identify and handle these missing values.

```python
df.isnull() # to check for missing values in the DataFrame
df.isnull().sum() # to count the number of missing values in each column
df.dropna() # to drop rows with missing values
df = df.dropna() # to drop rows with missing values and update the DataFrame
df.fillna(0) # to fill missing values with 0
```

## Filtering Data

Filtering allows us to select only the rows that satisfy a specific condition.

```python
df[df["Age"] > 20] # to filter rows where Age is greater than 20
df[df["Major"] == "CS"] # to filter rows where Major is "CS"
df[(df["Age"] > 20) & (df["Major"] == "CS")]#more than one condition
df[(df["Age"] > 20) | (df["Major"] == "CS")]
```

## Sorting Data

Sorting arranges the dataset in ascending or descending order.

```python
df.sort_values("Grade") # to sort the DataFrame by the Grade column in ascending order
df.sort_values("Grade", ascending=False) # to sort the DataFrame by the Grade column in descending order
df.sort_values(["Major", "Age"]) # to sort the DataFrame by Major and then by Age   
```

## Grouping Data

The `groupby()` function groups rows based on a column and performs calculations on each group.

```python
df.groupby("Major")["Grade"].mean() # to group the DataFrame by Major and calculate the mean Grade for each Major
df.groupby("Major").size() # to group the DataFrame by Major and count the number of students in each Major
df.groupby("Major")["Grade"].max() # to group the DataFrame by Major and find the maximum Grade for each Major
df.groupby("Major")["Grade"].min()
```
