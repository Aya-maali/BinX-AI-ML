## Task 1: Calculate Basic Statistics

Create a function that takes a list of numbers and returns the mean, minimum, and maximum values as a dictionary.

```python
def calculate_statistics(numbers):
    
    if not numbers:
        return None
    
    mean = sum(numbers) / len(numbers)
    smallest = min(numbers)
    largest = max(numbers)
    
    statistics = {
        'mean': mean,
        'min': smallest,
        'max': largest
    }
    return statistics

calculate_statistics([1, 2, 3, 4, 5])
    
```

## Task 2: Convert a for Loop to a List Comprehension

Rewrite a given for-loop that filters even numbers as a single list comprehension.

# given for-loop

for value in [1, 2, 3, 4]:
    if value % 2 == 0:
        print(f"{value} is even")
    else:
        print(f"{value} is odd")

```python
#single list comprehension.
even = [value for value in [1, 2, 3, 4] if value % 2 == 0]
print(even)
```

## Task 3: Define a small class representing a data record with at least two attributes and one method.

The class contains:
- Two attributes to store the record information.
- One method to display or process the stored data.

Classes help organize data and behavior together, which is important when working with data science libraries and machine learning frameworks.

```python
class Person:
    def __init__(self, name, age, city):
        self.name = name
        self.age = age
        self.city = city
    
    def display_info(self):
        print(f"Name: {self.name}, Age: {self.age}, City: {self.city}")

person1 = Person("Aya", 22, "Jenin")
person1.display_info()
```
