# Customer Churn Risk Prediction & Analysis

## Project Introduction

Customer churn is a common business problem where customers stop using a company's products or services.

The goal of this project is not only to predict whether a customer is likely to churn, but also to identify customers with a high churn risk and understand the main factors contributing to that risk.

The system will analyze customer information, predict the probability of churn, assign a risk level, identify important risk factors, and provide a possible retention action.

### Project Flow

```text
Customer Data
      ↓
EDA & Preprocessing
      ↓
Churn Prediction Model
      ↓
Churn Probability
      ↓
Risk Level
      ↓
Important Risk Factors
      ↓
Recommended Action
```

### Example Output

Customer: C1024

Churn Probability: 82%
Risk Level: HIGH

Main Risk Factors:
- Short contract
- Low engagement
- High monthly cost

Suggested Action:
Offer a retention plan

## Problem Statement

Customer churn can negatively affect a company's revenue and customer retention.

The main problem is identifying customers who are likely to leave before they actually churn.

This project aims to build a machine learning system that:

- Predicts the probability that a customer will churn.
- Classifies customers into different risk levels.
- Identifies the main factors contributing to the customer's churn risk.
- Provides a suggested retention action based on the identified risk.

The system will help transform customer data into actionable insights for customer retention.

## Project Objectives

The main objectives of this project are:

1. Understand the customer dataset through Exploratory Data Analysis (EDA).
2. Prepare and preprocess the data for machine learning.
3. Build a baseline churn prediction model.
4. Compare different machine learning models.
5. Evaluate the models using appropriate classification metrics.
6. Improve the selected model through preprocessing, feature engineering, and threshold tuning.
7. Estimate each customer's churn probability.
8. Assign customers to Low, Medium, or High risk levels.
9. Identify the main factors contributing to churn risk.
10. Deploy the final solution as a simple and usable application.

## Expected System Output

For each customer, the system should provide:

- Churn Probability
- Risk Level
- Churn Prediction
- Main Risk Factors
- Suggested Retention Action
